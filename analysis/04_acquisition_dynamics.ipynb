## The Paradox of Small Discounts

```python
import pandas as pd
from pathlib import Path

# Resolve paths relative to the project root regardless of CWD
base_path = Path(__file__).resolve().parent if '__file__' in dir() else Path.cwd()
base_path = base_path.parent if base_path.name in ('scripts', 'Personal_Applied', 'analysis') else base_path
GOLD_DIR = base_path / 'medallion' / 'gold'

# --- Customer base: gold_churn_features (12,819 retail customers) ---
# This excludes Bulk Import / wholesale customers, consistent with the
# churn and VTD analyses.
cf = pd.read_parquet(GOLD_DIR / 'gold_churn_features.parquet')

# --- Discount info: gold_discount_analysis (first orders, non-B2B) ---
da = pd.read_parquet(GOLD_DIR / 'gold_discount_analysis.parquet')
first_orders_da = da[
    (da['is_first_order'] == True) &
    (da['is_b2b_or_affiliate'] == False)
].drop_duplicates(subset=['customer_id'])

# --- Merge: start from CF base, left-join discount info ---
# Customers in CF but missing from DA will have NaN discount fields
# and will be classified as Full Price (no discount used).
first_orders = cf[['customer_id']].merge(
    first_orders_da,
    on='customer_id',
    how='left'
)

print(f'Customer base (aligned with churn analysis): {first_orders["customer_id"].nunique():,}')

# --- Comparison groups ---
first_orders['promo_group'] = 'Other/Small Discount'
first_orders.loc[first_orders['discount_type'].isna(), 'promo_group'] = 'Full Price'
first_orders.loc[first_orders['is_high_magnitude'] == True, 'promo_group'] = 'High Magnitude (30%+)'

# Numeric type fix
first_orders['repeat_purchase_90d'] = first_orders['repeat_purchase_90d'].astype(float)

analysis_1 = first_orders.groupby('promo_group').agg(
    new_customers=('customer_id', 'nunique'),
    repeat_rate_90d=('repeat_purchase_90d', 'mean')
).sort_values('repeat_rate_90d', ascending=False)

# Format repeat rate as percentage for slides
analysis_1['repeat_rate_90d'] = (analysis_1['repeat_rate_90d'] * 100).round(2).astype(str) + '%'

print('--- Analysis Part 1: The Repeat Rate Gap ---')
print(analysis_1)
```

> **Key Insight:** Our data shows that small, generic discounts produce the lowest 90-day retention (18.82%), performing worse than deep 30%+ discounts. This suggests that low-value incentives may attract the most price-sensitive, least-loyal customer segments.


## The Annualized Efficiency Gap

```python
# --- Analysis Part 2: ROI Translation ---

# calculate AOV (price_total is already in SGD — FX applied at silver layer)
avg_order_val = first_orders['price_total'].mean()

# defining 'Gap'
full_price_rate = first_orders[first_orders['promo_group'] == 'Full Price']['repeat_purchase_90d'].mean()
high_mag_rate = first_orders[first_orders['promo_group'] == 'High Magnitude (30%+)']['repeat_purchase_90d'].mean()
gap = full_price_rate - high_mag_rate
num_high_mag_customers = first_orders[first_orders['promo_group'] == 'High Magnitude (30%+)']['customer_id'].nunique()

# calculating 'lost potential revenue' 
# represents revenue LushProtein would have made if 738 customers behaved like normal loyal customers
lost_repeat_customers = round(num_high_mag_customers * gap)
lost_revenue = lost_repeat_customers * avg_order_val

print("--- Analysis Part 2: The Financial Impact ---")
print(f"Average Order Value: SGD {avg_order_val:.2f}")
print(f"Lost Repeat Customers (due to gap): {lost_repeat_customers}")
print(f"Estimated Lost Revenue per Cohort: SGD {lost_revenue:.2f}")
```

> **Key Insight:** The 1.58 percentage point retention gap translates to an estimated SGD 585 in lost repeat revenue per cohort cycle, annualising to approximately SGD 2,340. Shifting the promotional budget toward second-purchase re-engagement rather than front-end acquisition discounts represents a direct margin recovery opportunity.


## The Whale Effect: Average vs Median LTV Analysis

```python
# -- Analysis Part 3: LTV & Churn Risk --

# compare 12 month LTV; total revenue (already in SGD — FX applied at silver layer)
# answers for if high-magnitude discounts lead to lower overall spend over time
ltv_analysis = first_orders.groupby('promo_group').agg(
    avg_lifetime_value=('total_revenue', 'mean'),
    median_lifetime_value=('total_revenue', 'median')
).round(2)

# churn risk mapping
# % of customers that are 'dead' or 'at risk'
churn_segments = ['At Risk', 'Hibernating', 'About to Sleep']
first_orders['is_churn_risk'] = first_orders['rfm_group'].isin(churn_segments)

churn_analysis = first_orders.groupby('promo_group').agg(
    churn_risk_rate=('is_churn_risk', 'mean')
)
churn_analysis['churn_risk_rate'] = (churn_analysis['churn_risk_rate'] * 100).round(2).astype(str) + '%'

print("--- Analysis Part 3: LTV & Churn Risk ---")
print("\nLTV Comparison:")
print(ltv_analysis)
print("\nChurn Risk by Group:")
print(churn_analysis)
```

> **Key Insight:** Following FX normalisation, mean LTV stands at SGD 284.31 for organic buyers versus SGD 126.43 for high-magnitude discount buyers — a 2.25x gap. However, median LTV is comparable (SGD 68.82 vs SGD 77.86), confirming the divergence is driven by the concentration of high-spending Whale customers in the organic channel, not by differences in typical buyer behaviour.


### Data Quality Counter-measure: Corrected Churn Analysis
During analysis, we found that the `rfm_group` tags in the Gold layer were 100% null. To unblock the mid-term acquisition analysis, we are using a **"One-and-Done" proxy** (customers with only 1 total order) to accurately measure churn risk across discount tiers.

```python
# --- Analysis Part 3 (fixed): One-and-Done Churn ---
first_orders['is_one_and_done'] = first_orders['total_orders'] == 1

churn_fix = first_orders.groupby('promo_group').agg(
    one_and_done_rate=('is_one_and_done', 'mean')
)
churn_fix['one_and_done_rate'] = (churn_fix['one_and_done_rate'] * 100).round(2).astype(str) + '%'

print("--- Corrected Churn Analysis ---")
print(churn_fix)
```

> **Key Insight:** High-magnitude discount customers are **6.18 percentage points more likely** to churn (One-and-Done) compared to full-price customers (67.65% vs 61.48%). This suggests that deep discounting may be attracting price-sensitive shoppers rather than long-term brand loyalists.


```python
import numpy as np
import scipy.stats as stats

# observed frequencies based on data summary
# row 1: full-price/organic cohort  row 2: high-magnitude discount cohort
# columns: [did not repeat (one-and-done churned), repeated within 90 days]
full_price = first_orders[first_orders['promo_group'] == 'Full Price']
high_mag = first_orders[first_orders['promo_group'] == 'High Magnitude (30%+)']

fp_one_and_done = (full_price['total_orders'] == 1).sum()
fp_repeated = (full_price['total_orders'] > 1).sum()
hm_one_and_done = (high_mag['total_orders'] == 1).sum()
hm_repeated = (high_mag['total_orders'] > 1).sum()

observed_data = np.array([
    [fp_one_and_done, fp_repeated],
    [hm_one_and_done, hm_repeated]
])

print(f"Observed data:\n{observed_data}")

# execute the chi-square test of independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed_data)

print("="*50)
print("STATISTICAL HYPOTHESIS TEST: PROMO VS RETENTION")
print("="*50)
print(f"Chi-Square Statistic : {chi2_stat:.4f}")
print(f"P-Value              : {p_value:.4f}")
print(f"Degrees of Freedom   : {dof}")
print("="*50)

# interpret statistical significance
alpha = 0.05
if p_value < alpha:
    print("RESULT: Statistically Significant (p < 0.05).")
    print("We reject the Null Hypothesis. Front-end promotional discount magnitude")
    print("significantly alters long-term customer repeat behavior.")
else:
    print("RESULT: Not Statistically Significant.")
    print("Fail to reject the Null Hypothesis.")
print("="*50)
```

#### **Statistical Conclusion**
Because the $p\text{-value}$ is `0.0290` ($p < 0.05$), we reject the Null Hypothesis ($H_0$) with confidence. The relationship between front-end promotional incentives and heightened customer attrition is **statistically systemic**, not a random sample anomaly. Deep discounting successfully scales top-of-funnel acquisition volume while actively introducing high-churn user profiles into the system lifecycle.

#### **Key Insights & Strategic Recommendations**

* **The Customer Quality Illusion:** Looking strictly at **Average LTV**, organic buyers appear very much superior to deep discount shoppers ($SGD\ 284.31$ vs $SGD\ 126.43$). However, the **Median LTV** shows that the typical discount customer actually spends slightly *more* than a typical full-price customer ($SGD\ 77.86$ vs $SGD\ 68.82$). 
* **The 'Whale' Discovery Failure:** This distribution split proves that deep front-end discounts do not necessarily attract lower-spending everyday individuals; but that, **they completely fail to recruit the rare, high-value 'Whale' outliers** who drive the brand's core organic revenue. High-spending individuals are not motivated by standard acquisition discount codes.
* **Operational Recommendation (Inverting the Funnel):** We recommend protecting the front-end organic funnel to allow for high-value 'Whale' discovery. Deep promotional codes (30%+) should be phased out of acquisition marketing and deployed exclusively as a **re-engagement trigger for the second purchase** within the critical Day 30–60 consumption window.
